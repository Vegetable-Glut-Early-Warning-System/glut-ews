```python
# Crop-district footprint: supply-side rows exist only for these pairs
load("SELECT crop, district, status FROM core.dim_crop_district ORDER BY crop, status, district")
```

## First component-specific look

```python
# Real prices (CCPI-deflated): the first preprocessing step for glut detection
df["real_wholesale"] = df.wholesale_price_rs_kg / df.ccpi * 100
(df[df.economic_centre == "Peliyagoda"]
   .pivot_table(index="week_start_date", columns="crop", values="real_wholesale")
   .plot(figsize=(12, 4), title="Real wholesale price, Peliyagoda (Colombo)"));
```
